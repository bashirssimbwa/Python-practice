### Exercise 1: 
- Write a simple program to simulate the operation of the grep command on Unix. Ask the user to enter a regular expression and count the number of lines that matched the regular expression:

```python
import re
regex_input  = input("Enter a regular expression")
try:
    fhand = open('mbox-short.txt')
except FileNotFoundError:
    print("Unable to open File")
count = 0 
for line in fhand:
    line = line.rstrip()
    if re.search(regex_input, line):
        count +=1
print(f"mbox-short.txt had {count}  lines that matched {regex_input}")
        
```

### Exercise 2: 
- Write a program to look for lines of the form:

- New Revision: 39772
-  the number from each of the lines using a regular expression and the findall() method. Compute the average of the numbers and print out the average as an integer.

```python
import re
file_name = input("Enter File: ")
try:
    fhand = open(file_name)
except FileNotFoundError:
    print("File could not be opened: {file_name}")
total_revisions = 0
count = 0

for line in fhand:
    line = line.rstrip()
    matches = re.findall('^New Revision: ([0-9]+)', line)
    if len(matches) > 0:
        total_revisions += float(matches[0])
        count +=1
if count > 0:
    average = total_revisions /count
    print(int(average))
else:
    print("No matching revision lines found")
```
