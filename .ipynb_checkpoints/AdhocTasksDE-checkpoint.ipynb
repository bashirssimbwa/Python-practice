#### 0.1  Write a function that accepts a list of numbers and returns the minimum, maximum, total, count, and average  without using min(), max(), sum(), or statistics.

```python
def counter():
    count = 0
    total = 0
    max_value = None
    min_value = None
    average  = 0
    lst = []
    while True:
        inp = input("Enter a value or done")
        if not inp:
            print("Value cant be left blank")
            continue
        if inp.lower() == "done":
            break
        try:
            number = float(inp)
        except ValueError:
            print("Enter a valid Numerical Value")
            continue
        lst.append(number)
        count += 1
        total += number
        if count > 0:
            average = total / count
            for num in lst:
                if max_value is None or num > max_value:
                    max_value = num
                if min_value is None or num < min_value:
                    min_value = num 
    return lst, count, total, max_value, min_value, average
lst, count, total, max_value, min_value, average = counter()
print(f"FinalList: {lst}")
print(f"Total Count: {count}")
print(f"Total: {total}")
print(f"Maximum_value: {max_value}")
print(f"Minimum Value: {min_value}")
print(f"Average Value: {average}")
```

#### 0.2  Build a sentinel-loop program that reads transaction amounts until done. Reject non-numeric values, count valid transactions, and report total and average

```python
import decimal as Decimal
def amount():
    count = 0
    total = 0
    average  = 0
    max_value = None
    min_value = None
    lst = []
    while True:
        inp = input("Enter amount  or done to end transaction")
        if not inp:
            print("Input can be left blank")
            continue
        if inp.lower() == "done":
            break
        try:
            amount = float(inp)
            if amount < 0:
                print("Value cant be lees than Zero")
                continue
        except:
            print("Invalid input")
            continue
        count += 1
        lst.append(amount)
        if count > 0:
            total += amount
            average = total / count
            max_value = max(lst)
            min_value = min(lst)
    return lst, count , total, average, max_value, min_value
lst, count , total, average, max_value, min_value = amount()
print(f"amount Entered: {lst}")
print("========REPORT============")
print(f"Total Amount: R{total}")
print(f" Average Amount: R{average}")
```

### Question 3: Build afrequency conter for email addresses from a mail log Return the top sender and its count
Constraints: Ignore Lines that donnot begin with From :

```python
file = open("mbox-short.txt")
max_count = 0
min_count = 0
email_count = {}

for line in file:
    if line.startswith("From "):
        words = line.split()
        email = words[1]
        email_count[email] = email_count.get(email, 0) + 1
    for email, count in email_count.items():
        if max_count is None or count > max_count:
            max_email = email
            max_count = count
print(f"Most Committer: {max_email, max_count}")
```

### Question 4:  Given a list of dictionaries representing transactions, count transactions by currency and total the amount per currency.
- Skills: dict aggregation, nested data
- Constraints: Handle missing currency and non-numeric amount explicitly.
- Deliverable: Source code + a short note describing edge cases tested.

#### Step 1: 
- Generate the Dictionary of transactions

```python
import polars as pl
import numpy as np

num_rows = 2_000_0000

currencies = ["USD", "EUR", "GBP", "ZAR", "CAD"]
np.random.seed(42)

df = pl.DataFrame({
    "id": np.arange(1, num_rows + 1),
    "currency": np.random.choice(currencies, size=num_rows),
    "amount": np.round(np.random.uniform(5.0, 500.0, size=num_rows), 2)
})

print("Successfully created 20 million records!")
df.write_parquet("transactions.parquet")
print("Saved 20 Million records.parquet")

```

```python
transactions = pl.read_parquet("transactions.parquet")
# print(transactions[:3])

result = transactions.group_by("currency").agg(
    [
    pl.len().alias("transaction_count"),
    pl.col("amount").sum().round(2).alias("total_amount")   
    ]
)
print(result)
```
